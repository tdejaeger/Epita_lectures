# II) Combinations and permutations

```python

try:
    from math import perm, comb   # Python 3.8+
except ImportError:
    from math import factorial

    def perm(n, k):
        return factorial(n) // factorial(n - k)

    def comb(n, k):
        return factorial(n) // (factorial(k) * factorial(n - k))


```

## Exercise E — Cards: 5 hearts

- What is the probability of having 5 hearts if I pick 5 cards in a 52 card deck

- What is the probability of drawing 5 Hearts in a Specific Order from a 52 card-deck?(like A♥-2♥-3♥-4♥-5♥, or 5♥-6♥-7♥-8♥-9♥)

```python

# E1 unordered
outcome_E1=comb(13,5)
tot_E1=comb(52,5)
pE1 = outcome_E1/tot_E1
# E2 ordered specific sequence
outcome_E2=10*perm(5,5)
tot_E2=perm(52,5)
pE2 = outcome_E2/tot_E2
print('Probability of having 5 hearts is {:.3f}%.'.format(float(pE1)*100))
print('Probability of having 5 hearts in a specific order is {:.10f}%.'.format(float(pE2)*100))
```
