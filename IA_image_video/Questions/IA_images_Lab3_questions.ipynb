###**What is DataLoader?**

When working with neural networks, we rarely process one image (or data point) at a time.  
**Instead, we process data in “batches”**—small groups of samples—at each step of training or evaluation.

**Why use batches?**
- It’s much faster on modern hardware (GPUs love parallel computations).
- It makes learning more stable and efficient.

### The DataLoader: Efficient Batching

PyTorch’s `DataLoader` is the tool that handles batching for you.
It:
- Automatically splits your dataset into batches of the size you choose
- Shuffles your data if you want (for training)
- Loads data efficiently, even in parallel

**How it works:**
- When you use a `DataLoader`, each time you loop through it you get a `batch` (a tensor of multiple images and their corresponding labels).


### **13.2. DATA: Download and Prepare the MNIST Dataset**

```python
# ===========================================
# 2. DATA: Download and Prepare the MNIST Dataset
# ===========================================

# Step 1: Only convert to tensor (scales pixels to [0, 1])
mnist_dataset = datasets.MNIST('./data', train=True, download=True, transform=transforms.ToTensor())

# Load all training images
loader = DataLoader(mnist_dataset, batch_size=60000, shuffle=False)
images, labels = next(iter(loader))  # images: [60000, 1, 28, 28]

# Compute mean and std over all pixels
mean = images.mean().item()
std = images.std().item()
print(f"Mean: {mean:.4f}, Std: {std:.4f}")

#Most neural nets train better when input is roughly zero mean, unit variance.

transform = transforms.Compose([
    transforms.ToTensor(),  # Convert images to PyTorch tensors
    transforms.Normalize((mean,), (std,))  # Normalize to mean/std
])

batch_size = 128

trainset = datasets.MNIST('./data', train=True, download=True, transform=transform)
testset = datasets.MNIST('./data', train=False, download=True, transform=transform)

trainloader = torch.utils.data.DataLoader(trainset, batch_size=batch_size, shuffle=True)
testloader = torch.utils.data.DataLoader(testset, batch_size=batch_size, shuffle=False)

print(f"Training samples: {len(trainset)} | Test samples: {len(testset)}")

# Checking the dataset
for images, labels in trainloader:
    print('Image batch dimensions:', images.shape)
    print('Image label dimensions:', labels.shape)
    break
```

### **13.5 The model**

To create a neural network in PyTorch, you typically define a model class that inherits from nn.Module. This model class has two main parts:

Layer Definitions (__init__ method):

**1.   Layer Definitions** (__init__ method): Here you specify which layers your network will use—such as convolutional layers, pooling layers, flattening, and fully connected (linear) layers. These layers are the building blocks that will process your input data

**2.   Forward Pass** (forward method):
This method defines how the data flows through your model—in other words, how the input is transformed step by step into the output. You combine your defined layers and apply activation functions here. The forward method is called automatically when you pass data through the model.


- **nn.Module:** Base class for all models. You subclass it to create your own neural network.
- **nn.Conv2d:** Convolutional layer. Extracts spatial features from input images.
    - Key options: `in_channels`, `out_channels`, `kernel_size`, `stride`, `padding`
- **nn.MaxPool2d:** Downsamples the input by taking the maximum over a window (reduces spatial size, keeps strongest features). Can also be AveragePool.
- **nn.Flatten:** Flattens a tensor to (batch, all_features) for feeding into fully-connected layers.
- **nn.Linear:** Fully connected (dense) layer. Maps input features to output classes.
- **F.relu:** Non-linear activation function. Adds non-linearity, helps networks learn complex patterns (can be  Sigmoid, Tanh)
- **F.softmax:** Turns raw output scores (“logits”) into probabilities.


```python
# 5. CNN Model Definition

class ConvNet(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.conv_1 = nn.Conv2d(1, 8, 3, stride=1, padding=1)     # 28x28x1 -> 28x28x8
        self.pool_1 = nn.MaxPool2d(2, 2)                          # 28x28x8 -> 14x14x8
        self.flatten = nn.Flatten()
        self.linear_1 = nn.Linear(8 * 14 * 14, num_classes)

    def forward(self, x):
        x = self.pool_1(F.relu(self.conv_1(x)))
        x = self.flatten(x)
        logits = self.linear_1(x)
        probas = F.softmax(logits, dim=1)
        return logits, probas

# Model instantiation
num_classes = 10 #number 0 1 2 3 4 5 67 8 9
# Neural networks and data must be on the same device (CPU or GPU) for computation.
model = ConvNet(num_classes).to(device) #This moves the entire model’s parameters and buffers to the specified device
print(model)
```

## **13.5 Optimizers & Learning Rate**
When training a neural network, the optimizer is the algorithm that updates the model's weights to minimize the loss function. It decides how the model “learns” from its errors.

The most common optimizers are:

* SGD (Stochastic Gradient Descent): The classic and simplest optimizer. It updates weights in the direction that reduces the loss, using a small batch of data each time.

* Momentum (SGD with momentum):
An improved version of SGD that remembers the previous updates for smoother and faster convergence.

* Adam:
A very popular optimizer that adapts the learning rate for each parameter, often works well out of the box.

The learning rate (lr) is a critical parameter that controls how big the steps are when updating the model weights:

If the learning rate is too high, training may become unstable or miss the best solution.

If it's too low, training is very slow or can get stuck.

Typical values:

* SGD: lr=0.01 or 0.001 (but can vary)

* Adam: often lr=0.001 or 0.0001

### **13.7. Loss Function**

The **loss function** (also called the **cost function** or **criterion**) measures how well the model’s predictions match the true target labels. It quantifies the “error” made by the model, providing a feedback signal for learning.

The optimizer uses the loss value to update the model’s weights so that predictions improve over time.

**Common loss functions:**
- **CrossEntropyLoss:** For multi-class classification (e.g., MNIST digits).
- **MSELoss:** For regression (predicting continuous values).
- **BCELoss:** For binary classification (yes/no).

For most image classification problems (including this lab), we use **CrossEntropyLoss**, which combines log-softmax and negative log-likelihood in one step.


**Note:**  
- When you use `nn.CrossEntropyLoss`, you should **not** apply softmax to your model outputs.  
- `nn.CrossEntropyLoss` expects **raw logits** (the output before softmax) and the labels as integers (not one-hot).

### 13.8. Evaluation Function: Checking Model Accuracy

To evaluate our model during training or after, it’s helpful to have a reusable function that computes the **accuracy** on any dataset (training, validation, or test).

**Accuracy** is a good metric for classification when your data is balanced, as it simply measures the fraction of correctly predicted labels.

Below is a function that computes accuracy for any PyTorch model and dataloader.

```python
def compute_accuracy(model, data_loader):
    """
    Computes the accuracy (%) of the model on a given dataset.

    Args:
        model: the PyTorch model to evaluate
        data_loader: DataLoader for the dataset to check
    Returns:
        Accuracy in percent (%)
    """
    model.eval()  # Set model to evaluation mode
    correct_pred, num_examples = 0, 0

    with torch.no_grad():  # No gradients needed during evaluation
        for features, targets in data_loader:
            features = features.to(device)
            targets = targets.to(device)
            logits, probas = model(features)
            _, predicted_labels = torch.max(probas, 1)  # Class with highest probability
            num_examples += targets.size(0)
            correct_pred += (predicted_labels == targets).sum()

    accuracy = correct_pred.float() / num_examples * 100
    return accuracy.item()
```

To make our evaluation more robust, we can use **k-fold cross-validation**.  
This means splitting the training data into `k` parts (folds), training the model on (k-1) folds, and evaluating on the remaining fold, then averaging the results.

Below is a helper function using scikit-learn's `KFold` to split the data and report average accuracy.
python
Copier


```python
from sklearn.model_selection import KFold
import copy

def cross_validate_accuracy(model_class, dataset, k_folds=5, num_epochs=2, batch_size=64, lr=0.05):
    """
    Performs K-fold cross-validation and returns the average accuracy.
    Args:
        model_class: class of the model to create per fold (e.g. ConvNet)
        dataset: a torch Dataset (must support __getitem__ and __len__)
        k_folds: number of folds for cross-validation
        num_epochs: epochs to train per fold
        batch_size: batch size for loaders
        lr: learning rate
    Returns:
        Average validation accuracy over k folds
    """
    kf = KFold(n_splits=k_folds, shuffle=True, random_state=42)
    acc_list = []
    indices = list(range(len(dataset)))

    for fold, (train_idx, val_idx) in enumerate(kf.split(indices)):
        # Build dataloaders for this fold
        train_sampler = torch.utils.data.Subset(dataset, train_idx)
        val_sampler = torch.utils.data.Subset(dataset, val_idx)
        train_loader = torch.utils.data.DataLoader(train_sampler, batch_size=batch_size, shuffle=True)
        val_loader = torch.utils.data.DataLoader(val_sampler, batch_size=batch_size)

        # New model for each fold
        model = model_class().to(device)
        optimizer = torch.optim.SGD(model.parameters(), lr=lr)
        criterion = nn.CrossEntropyLoss()

        # Train for a few epochs
        for epoch in range(num_epochs):
            model.train()
            for images, labels in train_loader:
                images, labels = images.to(device), labels.to(device)
                optimizer.zero_grad()
                logits, probas = model(images)
                loss = criterion(logits, labels)
                loss.backward()
                optimizer.step()

        # Evaluate on validation fold
        acc = compute_accuracy(model, val_loader)
        acc_list.append(acc)
        print(f"Fold {fold+1}/{k_folds} accuracy: {acc:.2f}%")

    print(f"\nAverage {k_folds}-fold validation accuracy: {np.mean(acc_list):.2f}%")
    return acc_list
```

### **13.9. Training the Model**
Now it’s time to train the model!  
The training loop will:
- Pass batches of images through the model,
- Compute the loss (how wrong the model is),
- Update the model’s weights using the optimizer.

We’ll also print the loss and accuracy every epoch to track progress.


```python
num_epochs = 5  # You can adjust this (more epochs for better results)

for epoch in range(num_epochs):
    model.train()  # Set the model to training mode
    running_loss = 0.0 #Accumulate loss to average later
    running_corrects = 0
    total_examples = 0

    for images, labels in trainloader:
        images = images.to(device)
        labels = labels.to(device)

        # Forward pass: Compute predictions and loss
        #Data goes through the model to get predictions (logits = raw scores; probas = probabilities)
        logits, probas = model(images)
        #Computes how wrong the model is

        loss = criterion(logits, labels)

        # Backpropagation and optimization
        optimizer.zero_grad() # Clear previous gradients
        loss.backward() #Computes gradients (how to change the weights to reduce loss)

        ### UPDATE MODEL PARAMETERS

        optimizer.step() #Updates the model’s weights in the direction that reduces the loss.


        # Statistics
        running_loss += loss.item() * images.size(0)
        _, preds = torch.max(probas, 1)
        running_corrects += (preds == labels).sum().item()
        total_examples += labels.size(0)

    # Calculate average loss and accuracy for the epoch
    epoch_loss = running_loss / total_examples
    epoch_acc = running_corrects / total_examples * 100

    print(f"Epoch {epoch+1}/{num_epochs}: Loss = {epoch_loss:.4f}, Accuracy = {epoch_acc:.2f}%")

    # (Optional) Evaluate on test set after each epoch
    test_acc = compute_accuracy(model, testloader)
    print(f"Test Accuracy after epoch {epoch+1}: {test_acc:.2f}%")
```

### **13.11. Hyperparameter tuning**

* Number of epochs (how many times the model sees the training data)

* Optimizer type (e.g., Adam, SGD)

### **Test new models**

3 convolutional blocks, each with 2× Conv2d + BatchNorm + ReLU (so 6 convolutional layers in total), plus Dropout in conv and FC blocks.

```python
class CifarCNN_new(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        # --- First conv block ---
        self.conv1_1 = nn.Conv2d(3, 32, kernel_size=3, padding=1)    # 32x32x3 -> 32x32x32
        self.bn1_1   = nn.BatchNorm2d(32)
        self.conv1_2 = nn.Conv2d(32, 64, kernel_size=3, padding=1)   # 32x32x32 -> 32x32x64
        self.bn1_2   = nn.BatchNorm2d(64)
        self.pool1   = nn.MaxPool2d(2, 2)                            # 32x32x64 -> 16x16x64

        # --- Second conv block ---
        self.conv2_1 = nn.Conv2d(64, 128, kernel_size=3, padding=1)  # 16x16x64 -> 16x16x128
        self.bn2_1   = nn.BatchNorm2d(128)
        self.conv2_2 = nn.Conv2d(128, 128, kernel_size=3, padding=1) # 16x16x128 -> 16x16x128
        self.bn2_2   = nn.BatchNorm2d(128)
        self.pool2   = nn.MaxPool2d(2, 2)                            # 16x16x128 -> 8x8x128
        self.drop2d2 = nn.Dropout2d(p=0.05)

        # --- Third conv block ---
        self.conv3_1 = nn.Conv2d(128, 256, kernel_size=3, padding=1) # 8x8x128 -> 8x8x256
        self.bn3_1   = nn.BatchNorm2d(256)
        self.conv3_2 = nn.Conv2d(256, 256, kernel_size=3, padding=1) # 8x8x256 -> 8x8x256
        self.bn3_2   = nn.BatchNorm2d(256)
        self.pool3   = nn.MaxPool2d(2, 2)                            # 8x8x256 -> 4x4x256

        # --- Fully connected layers ---
        self.drop_fc1 = nn.Dropout(p=0.1)
        self.linear_1 = nn.Linear(256*4*4, 1024)                     # 4096 -> 1024
        self.linear_2 = nn.Linear(1024, 512)                         # 1024 -> 512
        self.drop_fc2 = nn.Dropout(p=0.1)
        self.linear_3 = nn.Linear(512, num_classes)                  # 512 -> 10

    def forward(self, x):
        # --- First conv block ---
        x = F.relu(self.bn1_1(self.conv1_1(x)))       # (B, 32, 32, 32)
        x = F.relu(self.bn1_2(self.conv1_2(x)))       # (B, 64, 32, 32)
        x = self.pool1(x)                             # (B, 64, 16, 16)

        # --- Second conv block ---
        x = F.relu(self.bn2_1(self.conv2_1(x)))       # (B, 128, 16, 16)
        x = F.relu(self.bn2_2(self.conv2_2(x)))       # (B, 128, 16, 16)
        x = self.pool2(x)                             # (B, 128, 8, 8)
        x = self.drop2d2(x)                           # (B, 128, 8, 8)

        # --- Third conv block ---
        x = F.relu(self.bn3_1(self.conv3_1(x)))       # (B, 256, 8, 8)
        x = F.relu(self.bn3_2(self.conv3_2(x)))       # (B, 256, 8, 8)
        x = self.pool3(x)                             # (B, 256, 4, 4)

        # --- Flatten ---
        x = x.view(x.size(0), -1)                     # (B, 4096)

        # --- Fully connected ---
        x = self.drop_fc1(x)
        x = F.relu(self.linear_1(x))                  # (B, 1024)
        x = F.relu(self.linear_2(x))                  # (B, 512)
        x = self.drop_fc2(x)
        logits = self.linear_3(x)                     # (B, num_classes)
        probas = F.softmax(logits, dim=1)
        return logits, probas
```

```python
# loss criterion
criterion = nn.CrossEntropyLoss()

model = CifarCNN_new(10).to(device) # no need for parameters as we alredy defined them in the class
# optimizer - SGD, Adam, RMSPr
learning_rate = 1e-4
optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)
```
